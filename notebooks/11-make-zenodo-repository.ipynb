# 🏛️ Make a `zenodo` Repository

This notebook collects important data files that we should publish along with the paper. These include:

- the posterior from the `no-magma` baseline fit, for use in predicting atmosphere probabilities 
- for the subsets `['all', 'solar', 'exo']` and the atmosphere fluxlimits `['any', 'no-magma', 'no-magma-no-freeze]`:
    - the organized exoatlas populations including assumed planet properties 
    - the posterior samples for a shoreline fit to each specific subset (= exoplanet or Solar System or both) and fluxlimit (= applied flux limits), with and without uncertainties on the planet parameters 
- additional figures and animations beyond those in the paper, including versions with and without planet annotations


```python
!rm -rfv upload-to-zenodo* 
!mkdir upload-to-zenodo
!cp cosmic-shoreline-btwm2026-posterior.nc upload-to-zenodo/.
```

```python
from shoreline import * 

directories = ['posteriors', 'organized-exoatlas-populations', 'figures']
base_directory = 'upload-to-zenodo'

mkdir(base_directory)
for d in directories:
    mkdir(f'{base_directory}/{d}')
    if d == 'figures':
        for f in glob.glob(f'{d}/*{suffix}*'):
            shutil.copy(f, f'{base_directory}/{d}/.')
    else:
        for fluxlimit in ['any', 'no-magma', 'no-magma-no-freeze']:
            for subset in ['all', 'exo', 'solar']:
                suffix = f'{subset}+{fluxlimit}'
                mkdir(f'{base_directory}/{d}/{fluxlimit}/')
                for f in glob.glob(f'{d}/*{suffix}*'):
                    shutil.copy(f, f'{base_directory}/{d}/{fluxlimit}/.')
    shutil.make_archive(f'{base_directory}/{d}', 'zip', f'{base_directory}/{d}/')
    shutil.rmtree(f'{base_directory}/{d}/')
```
